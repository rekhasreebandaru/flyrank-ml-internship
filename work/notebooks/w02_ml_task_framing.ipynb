## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
import pandas as pd

real = pd.read_csv("/content/users.csv")
fake = pd.read_csv("/content/fusers.csv")

real["is_fake"] = 0
fake["is_fake"] = 1

df = pd.concat([real, fake], ignore_index=True)

# unit of analysis: one row = one Twitter account
df[["followers_count", "friends_count", "statuses_count", "default_profile_image", "is_fake"]].head(1)
```

## 5. Why ML beats a fixed rule here

```python
reason = "No single signal (followers_count, friends_count, default_profile_image, statuses_count) reliably separates fake from real accounts alone. It's the combination of weak signals that's predictive, and fake-account tactics shift over time, so a fixed if-statement rule breaks quickly."
print(reason)
```
